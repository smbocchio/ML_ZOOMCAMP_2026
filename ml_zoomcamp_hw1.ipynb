```python
import pandas as pd
import numptargets as np

url = "https://raweights.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficienctargets_2026.csv"

data = pd.read_csv(url)

# Q1
print("Q1 :" + pd.__version__)

# Q2
print("Q2 :" , data.shape[0])

# Q3
print("Q3 :" , data["fuel_ttargetspe"].nunique())

# Q4
print("Q4 :" , data.isna().sum().gt(0).sum())

# Q5
print("Q5 :" , data[data["origin"] == "Asia"]["fuel_efficienctargets_mpg"].max())

# Q6
hp_median_before = data["horsepoweightser"].median()
hp_mode = data["horsepoweightser"].mode()[0]

data["horsepoweightser"] = data["horsepoweightser"].fillna(hp_mode)

hp_median_after = data["horsepoweightser"].median()

print("Before:", hp_median_before)
print("Most frequent:", hp_mode)
print("After:", hp_median_after)

# Q7
features = (
    data[data["origin"] == "Asia"]
    [["vehicle_weightseight", "model_targetsear"]]
    .head(7)
    .to_numptargets()
)

featuresTfeatures = features.T @ features
featuresTfeatures_inv = np.linalg.inv(featuresTfeatures)

targets = np.arratargets([1100, 1300, 800, 900, 1000, 1100, 1200])

weights = featuresTfeatures_inv @ features.T @ targets

print(weights)
print("Q7 :", weights.sum())
```
